## Basic Data Types in Python 

Before looking at specific data types, it helps to be clear on what a **variable** actually is. A variable is simply a name that points to a value stored in memory. When you write `share_price = 42.35`, Python creates the value `42.35`, stores it in memory, and binds the name `share_price` to it — from then on, using `share_price` in your code refers to that value.

Unlike some other programming languages, Python does not require you to declare a type upfront: a variable simply takes on the type of whatever value is assigned to it, and it can even be reassigned to a value of a different type later on (this is called *dynamic typing*).

Every value in Python has a type, and the type determines which operations are valid — you can add two `int`s together, but not an `int` and a `str`. We can check the type of any value with the built-in `type()` function. Let's look at the most important basic types: `int`, `float`, `str`, and `bool`.

#### `int` — Whole Numbers
The `int` type represents whole numbers, positive or negative, with no decimal point. In finance, an `int` is a natural choice for counts such as a number of shares or a calendar year.

````python
# int - whole numbers
num_shares = 100
founding_year = 2015

print(num_shares, type(num_shares))
print(founding_year, type(founding_year))
````

#### `float` — Decimal Numbers
The `float` type represents numbers with a decimal point, used for anything measured on a continuous scale — such as a share price or an interest rate. Recall from the earlier arithmetic example that dividing two `int`s (e.g. `8 / 2`) also produces a `float` in Python.

````python
# float - decimal numbers
share_price = 42.35
interest_rate = 0.025

print(share_price, type(share_price))
print(interest_rate, type(interest_rate))

# Dividing two ints returns a float
print(8 / 2, type(8 / 2))
````

#### `str` — Text
The `str` (string) type represents text, written between single or double quotes. Strings are commonly used for identifiers such as a company name or a ticker symbol.

````python
# str - text
ticker = "NESN"
company_name = "Nestlé"

print(ticker, type(ticker))
print(f"{company_name} ({ticker})")
````

#### `bool` — True or False
The `bool` type has only two possible values: `True` and `False`. While you can assign a boolean directly, booleans most often arise as the *result* of a comparison, such as checking whether a share price exceeds a threshold.

````python
# bool - True or False
is_dividend_paying = True
print(is_dividend_paying, type(is_dividend_paying))

# Booleans often come from comparisons
print(share_price < 40, type(share_price < 40))
````

#### Checking and Converting Types
We can always check a value's type with `type()`, and convert (or *cast*) a value from one type to another using the `int()`, `float()`, and `str()` functions. This is especially useful when reading data from external sources such as a CSV file, where everything typically arrives as text.

````python
# Converting a numeric string to a float works fine
price_str = "42.35"
price_float = float(price_str)
print(price_float, type(price_float))

# But converting the same string directly to an int does not:
# int(price_str)
````

The error above (`ValueError`) occurs because `"42.35"` is not a whole number, so Python cannot interpret it as an `int` directly. To get an integer, first convert to `float`, then to `int` (which truncates the decimal part):

````python
price_int = int(float(price_str))
print(price_int, type(price_int))
````

#### Compound Types
So far, each variable has held a single value. Python also has *collection* types that hold multiple values in one variable:

- **`set`** — an unordered collection of unique values
- **`tuple`** — an ordered, *unchangeable* collection
- **`list`** — an ordered, changeable collection (covered in detail next)
- **`dict`** — a collection of key–value pairs

````python
coordinates = (47.36, 8.55)              # tuple: fixed pair of coordinates
fx_rates = {"USD": 0.91, "GBP": 0.79}    # dict: currency -> exchange rate
currencies = {"USD", "EUR", "GBP"}       # set: unique currency codes

print(coordinates, type(coordinates))
print(fx_rates, type(fx_rates))
print(currencies, type(currencies))
````

#### Lists
A list is one of the most fundamental and flexible data structures in Python. It allows you to store multiple values in a single variable, making it essential for handling datasets, computations, and iterative tasks. Lists can contain numbers, strings, or even other lists, and they support operations like indexing, slicing, sorting, and appending new elements. Unlike arrays in some other programming languages, Python lists are dynamic, meaning they can grow or shrink as needed. 

````python
# Creating a list of numbers
numbers = [53, 50, 45, 70, 62, 56, 80]
````

````python
# Accessing elements
print(numbers[0])  
print(numbers[-1]) 
````

````python
# Sorting the list
numbers.sort()

print(numbers) 
````

````python
# Calculate Mean (Average)
import numpy as np
mean_value = np.mean(numbers)
print(f"Mean (Average): {mean_value}")
````

````python
# Calculate Median
median_value = np.median(numbers)
print(f"Median: {median_value}")
````

````python
# Calculate First Quartile (Q1) and Third Quartile (Q3)
q1 = np.percentile(numbers, 25)  # 25th percentile (Q1)
q3 = np.percentile(numbers, 75)  # 75th percentile (Q3)
print(f"First Quartile (Q1): {q1:.2f}")
print(f"Third Quartile (Q3): {q3:.2f}")
````

````python
# Calculate Sample Variance (unbiased estimator)
sample_variance = np.var(numbers, ddof=1)  # ddof=1 ensures sample variance
print(f"Sample Variance: {sample_variance}")
````

#### Dictionaries
A dictionary is another fundamental Python data structure. Unlike a list, which stores values by their position (index), a dictionary stores values by a unique **key** — making it ideal for lookups, such as finding a stock's price by its ticker symbol. Dictionaries are written with curly braces `{}` and `key: value` pairs.

````python
# Creating a dictionary of stock prices by ticker
stock_prices = {"AAPL": 325.13, "MSFT": 501.02, "NESN": 88.40, "UBSG": 27.55}
print(stock_prices)
````

#### Accessing and Updating Values
Access a value with its key using square brackets — similar to a list, but with a key instead of a position. Assigning to a new key adds it to the dictionary; assigning to an existing key updates its value.

````python
# Accessing a value by its key
print(stock_prices["AAPL"])

# Adding a new key-value pair
stock_prices["GOOGL"] = 178.32

# Updating an existing value
stock_prices["AAPL"] = 330.00

print(stock_prices)
````

#### Checking for a Key
Looking up a key that doesn't exist raises a `KeyError`:

````python
#stock_prices["TSLA"]
````

The error above (`KeyError`) occurs because `"TSLA"` isn't in the dictionary. To avoid this, check for the key first with the `in` operator, or use `.get()` to supply a default value instead of raising an error:

````python
# Check membership first with 'in'
print("TSLA" in stock_prices)

# .get() returns a default value instead of raising an error
price = stock_prices.get("TSLA", "Not available")
print(price)
````

#### Iterating Over a Dictionary
Loop over a dictionary's keys, values, or key-value pairs with `.keys()`, `.values()`, and `.items()`:

````python
# Iterating over key-value pairs
for ticker, price in stock_prices.items():
    print(f"{ticker}: {price}")

# Calculating the average price across all stocks
import numpy as np
average_price = np.mean(list(stock_prices.values()))
print(f"Average Price: {average_price:.2f}")
````

## 1. Data Analysis and Visualization:

Python provides powerful libraries for both data analysis and visualization, making it an ideal tool for empirical research. Among these, Matplotlib and Seaborn are two essential libraries for creating high-quality visualizations. Matplotlib is a versatile library that allows users to create a wide range of static, animated, and interactive plots, making it the foundation of Python’s visualization ecosystem. On top of that, Seaborn builds on Matplotlib, offering a more intuitive interface and beautiful default styles for statistical visualizations, such as histograms, box plots, and regression plots.

````python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import statsmodels.api as sm

# Set Seaborn style for better aesthetics
sns.set_style("darkgrid")

# Generate a sample dataset
np.random.seed(42)  # For reproducibility
data = pd.DataFrame({
    "x": np.linspace(1, 100, 100),  # A simple range
    "y": np.linspace(1, 100, 100) + np.random.normal(0, 10, 100),  # Adding some noise
    "category": np.random.choice(["A", "B", "C"], 100)  # Categorical variable
})
# Display the first few rows of the dataset
#print(data.head())
data.head()
````

````python

## ---- Matplotlib Example: Basic Line Plot ---- ##
plt.figure(figsize=(8,5))
plt.plot(data["x"], data["y"], color="blue", linestyle="--", marker="o", label="Noisy Linear Data")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Matplotlib: Basic Line Plot Example")
plt.legend()
plt.show()

## ---- Seaborn Example: Scatter Plot with Regression Line ---- ##
plt.figure(figsize=(8,5))
sns.regplot(x="x", y="y", data=data, scatter_kws={"color": "darkblue"}, line_kws={"color": "red"})
plt.title("Seaborn: Regression Line with Scatter Plot")
plt.show()

## ---- Seaborn Example: Box Plot ---- ##
plt.figure(figsize=(8,5))
sns.boxplot(x="category", y="y", data=data, palette="coolwarm")
plt.title("Seaborn: Box Plot of Y by Category")
plt.show()

## ---- Seaborn Example: Distribution Plot ---- ##
plt.figure(figsize=(8,5))
sns.histplot(data["y"], kde=True, bins=20, color="green")
plt.title("Seaborn: Distribution of Y with KDE")
plt.xlabel("Y values")
plt.show()
````

## 2. Statistical and Econometric Modeling 

Statistical and econometric modeling relies on random numbers, probability distributions, and probability calculations to analyze real-world data and test economic theories. In empirical research, we often work with random variables that follow specific probability distributions, such as the normal distribution (e.g., modeling stock returns), the Poisson distribution (e.g., modeling the number of customer arrivals), or the binomial distribution (e.g., modeling success/failure outcomes). These distributions help us simulate real-world scenarios, generate synthetic data, and quantify uncertainty. Using Python, we can draw random samples from these distributions and compute key probabilities. For instance, we can answer questions like: What is the probability that a stock return is below a certain threshold? or How likely is it that a firm will experience a given number of customer visits in an hour? Python's numpy and scipy.stats libraries provide powerful tools to generate random numbers, calculate probabilities, and estimate statistical properties, making it an essential toolkit for econometric modeling and hypothesis testing. Let’s explore some examples in Python! 

````python
# 1. Generate random numbers from a normal distribution (mean=0, std=1)
normal_data = np.random.normal(loc=0, scale=1, size=1000)
````

````python
# 2a. Probability of a value being <= 1.5 in a standard normal distribution
prob_a = stats.norm.cdf(1.5)  
print(f"Probability P(X <= 1.5) in N(0,1): {prob_a}")

# 2b. Probability of a value being >= 2 in a standard normal distribution
prob_b = 1 - stats.norm.cdf(2) 
print(f"Probability P(X >= 2) in N(0,1): {prob_b}")

# 2c. Probability of a value being <= 1.5 but also >= 2 in a standard normal distribution
prob_c = stats.norm.cdf(2) - stats.norm.cdf(1.5) 
print(f"Probability P(1.5 <= X <= 2) in N(0,1): {prob_c}")
````

````python
# 3. Visualizing probability distributions
x = np.linspace(-4, 4, 1000)
plt.plot(x, stats.norm.pdf(x), label="Normal Distribution (μ=0, σ=1)", color='blue')
plt.xlabel("X values")
plt.ylabel("Probability Density")
plt.title("Probability Density Function of a Normal Distribution")
plt.legend()
plt.show()
````

#### Relationships Between Variables
Beyond looking at variables individually, we often want to know how they relate to one another — for example, does age correlate with attitudes toward photovoltaic energy policy? Plotting one variable against another (e.g. a boxplot of age across response categories) gives a first visual impression before we quantify the relationship formally.

````python
# Step 1: map the text labels to their integer codes
label_map = {
    'Sicher nein': 1,
    'Eher nein': 2,
    'Eher ja': 3,
    'Sicher ja': 4,
    'Weiss nicht': -8,
}

df_env['endorses_photovoltaic'] = df_env['photovoltaic'].map(label_map)

# Step 2: invert the 1-4 scale (1<->4, 2<->3) wherever the question was framed
# as a ban ("Verbietet"), but leave -8 ("don't know") untouched
is_forbid = df_env['photovoltaic_treat2'] == 'Verbietet'
is_valid_scale = df_env['endorses_photovoltaic'].isin([1, 2, 3, 4])

df_env.loc[is_forbid & is_valid_scale, 'endorses_photovoltaic'] = (
    5 - df_env.loc[is_forbid & is_valid_scale, 'endorses_photovoltaic']
)

df_env['endorses_photovoltaic'] = df_env['endorses_photovoltaic'].astype('Int64')  # capital I = nullable integer
print(df_env['photovoltaic'].isna().sum())
print(df_env['endorses_photovoltaic'].isna().sum()) # NaN include former -99
````

### Linear Regression
A linear regression lets us go beyond a visual impression and quantify the relationship between variables: how much does the outcome change, on average, for a one-unit change in a predictor — and is that relationship statistically distinguishable from zero? We'll use the statsmodels library, which provides detailed regression output (coefficients, p-values, R²) similar to what you'd see in Stata or R.

Note: our outcome here is a 4-point ordinal scale rather than a truly continuous variable. Treating it as continuous in an OLS regression is a common simplification for a first pass — the methodologically stricter approach would be an ordinal logistic regression, which we mention but don't cover here.

#### Define the SLR model
We want to fit the model: 

$$ endorses\_photovoltaic_i = \beta_0 + \beta_1 age_i + u_i$$


````python
import statsmodels.api as sm

X = sm.add_constant(reg_data['age'])
y = reg_data['endorses_photovoltaic'].astype(float)

model = sm.OLS(y, X).fit()
print(model.summary())
````

#### Expand simple model to Multiple Linear Regression (MLR)
We want to fit the model: 

$$ endorses\_photovoltaic_i = \beta_0 + \beta_1 age_i + \beta_2 female_i + \beta_3 treat\_verbietet_i + \beta_4 treat\_Kanton_i + u_i$$


````python
reg_data = reg_data.copy()
reg_data['female'] = df_env['female']
reg_data['treat_verbietet'] = (df_env['photovoltaic_treat2'] == 'Verbietet').astype(int)
reg_data['treat_kanton'] = (df_env['photovoltaic_treat1'] == 'Kanton').astype(int)


X = sm.add_constant(reg_data[['age', 'female', 'treat_verbietet', 'treat_kanton']])
y = reg_data['endorses_photovoltaic'].astype(float)

model = sm.OLS(y, X).fit()
print(model.summary())
````
